6.	Given a 'marks' variable, print the grade:
- 90 and above: 'A'
- 75–89: 'B'
- 50–74: 'C'
- Below 50: 'Fail'


```python
marks = 53

if marks >= 90:
    print('A')
elif marks >= 75 and marks <= 89:
    print('B')
elif marks >= 50 and marks <= 74:
    print('C')
else:
    print('Fail')
```

7.	Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
num = 28

if num > 0:
    if num % 2 == 0:
        print("Positive and Even Number")
    else:
        print("Positive but Odd Number")
elif num == 0:
    print("Number is Zero")
else:
    print("Negative Number")
```

11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
dataframe = pd.DataFrame([['Abhishek',32,'Hyderabad',120000],
                 ['Head',35,'Nizambad',130000],
                 ['Ishan',30,'Khammam',100000],
                 ['Klassen',40,'Jayithal',125000],
                 ['Reddy',25,'Vizag',100500],
                 ['Abhishek',32,'Hyderabad',120000],
                 ['Ankiet varma',38,'Rangareddy'],
                 ['Pat Cummins',45,'Wrangal',200000],
                 ['Shivang',30,'Suryapet'],
                 ['Klassen',40,'Jayithal',125000]],columns = ['Name','Age','City','Salary'])       # creating a dataframe with 10 rows and 4 columns
```

```python
print(dataframe)
```

```python
dataframe.info()
```

```python
dataframe.describe()
```

```python
print(dataframe[['Name','City']])  # Select the city and name columns
```

```python
df2 = dataframe.drop('City',axis=1)   # Drop city column
```

```python
print(df2)
```

```python
df2.isnull().sum()  # Finding null values
```

```python
df2['Salary'].mean()
```

```python
df2['Salary'] = df2['Salary'].fillna(df2['Salary'].mean())
print(df2)                #  
```
