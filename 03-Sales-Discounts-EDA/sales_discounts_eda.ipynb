## Data Visualization

### Histograms:
	Plot histograms for each numerical column.


```python
df['Volume'].hist()
plt.title('Volume')
plt.show()
```

Most of the volume values are concentrated between 1 and 6. The histogram is positively skewed (right-skewed) because there are a few high-volume values on the right side. This indicates that low-volume sales occur more frequently, while high-volume sales are less common. A few extreme values may be present in the dataset.

```python
df['Avg Price'].hist()
plt.title('Avg Price')
plt.show()
```

Most products have lower average prices, while only a few products have very high prices. The distribution is right-skewed because the data extends towards higher price values. Some high-priced products may act as outliers.

```python
df['Total Sales Value'].hist()
plt.title('Total Sales Value')
plt.show()
```

Most sales transactions have lower sales values, and only a few transactions have very high sales values. The distribution is positively skewed. This indicates that high-value sales occur less frequently.

```python
df['Discount Rate (%)'].hist()
plt.title('Discount Rate (%)')
plt.show()
```

The discount rates are mainly concentrated between 15% and 20%. The distribution is not perfectly symmetric and shows variation in discount percentages. There are no major outliers visible in the histogram.

```python
df['Discount Amount'].hist()
plt.title('Discount Amount')
plt.show()
```

Most discount amounts are relatively low, while a few transactions have very high discounts. The histogram is right-skewed with a long tail towards higher values. This suggests the presence of some large discount transactions.

```python
df['Net Sales Value'].hist()
plt.title('Net Sales Value')
plt.show()
```

Most net sales values are concentrated at lower ranges, while a few transactions have very high net sales values. The distribution is positively skewed. Some high-value transactions may be considered outliers.

## Box Plots:

```python
df['Volume'].plot(kind='box')
plt.title('Volume')
plt.show()
```

The boxplot shows several outliers above the upper whisker, indicating some unusually high volume values. The IQR is relatively small, which means most volume values are concentrated within a narrow range.

```python
df['Avg Price'].plot(kind='box')
plt.title('Avg Price')
plt.show()
```

The boxplot shows a few high-price outliers above the upper whisker. The IQR indicates that most products have prices within a limited range, while a small number of products have exceptionally high prices.

```python
df['Total Sales Value'].plot(kind='box')
plt.title('Total Sales Value')
plt.show()
```

Several outliers are present in the higher sales value range. The IQR is moderate, showing that the middle 50% of sales values are concentrated within a specific range, while a few transactions have very high sales values.

```python
df['Discount Rate (%)'].plot(kind='box')
plt.title('Discount Rate (%)')
plt.show()
```

The boxplot shows some outliers in the lower discount rate range. The IQR is relatively small, indicating that most discount rates are concentrated between similar values. Overall, the discount rates are fairly consistent.

```python
df['Discount Amount'].plot(kind='box')
plt.title('Discount Amount')
plt.show()
```

Many outliers are visible above the upper whisker, indicating a few transactions received very high discounts. The IQR shows that most discount amounts are concentrated in a lower range, while some extreme values are present.

```python
df['Net Sales Value'].plot(kind='box')
plt.title('Net Sales Value')
plt.show()
```

The boxplot shows several high-value outliers above the upper whisker. The IQR indicates that the middle 50% of net sales values are concentrated within a specific range, while a few transactions have exceptionally high net sales values.

## Bar Chart Analysis for Categorical Column:

```python
df.select_dtypes(include='object').columns  # 	Identify categorical columns in the dataset.
```

```python
df['BU'].value_counts().plot(kind='bar')
plt.title('BU')
plt.show()
```

The three business units (FMCG, Mobiles, and Lifestyle) have almost the same number of records. This shows that the dataset is evenly distributed across all business units.

```python
df['Day'].value_counts().plot(kind='bar')
plt.title('Day')
plt.show()
```

Thursday has the highest number of records compared to other days. The remaining days have almost equal frequencies in the dataset.

```python
df['Brand'].value_counts().plot(kind='bar')
plt.title('Brand')
plt.show()
```

Jeera has the highest number of records in the dataset. Gear and Viva follow next, while Sumsang and Orange have the lowest frequencies. This indicates that some brands are more represented than others.

#### Other categorical columns such as City, SKU, and Model were examined, but they showed either uniform distribution or a single category, providing limited analytical insight.
